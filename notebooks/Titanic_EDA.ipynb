```python
# Identify columns that contain missing values

columns_with_missing = df.columns[df.isnull().any()]

print("Columns Requiring Missing Value Treatment:")
print(columns_with_missing)
```

### Columns Requiring Missing Value Treatment

The missing-value analysis shows that the `Age`, `Cabin`, and `Embarked` columns contain missing values. These columns require further consideration before proceeding with data cleaning. The percentage of missing values will be used to determine the most appropriate treatment strategy in the next phase.

### Missing Value Treatment Strategy

For this dataset, `fillna()` will be used instead of `dropna()` because removing rows with missing values could unnecessarily reduce the number of passenger records.

- `Age` will be filled using the median because it is a numerical feature and the median is less sensitive to extreme values.
- `Embarked` will be filled using the mode because it is a categorical feature with only a small number of missing values.
- `Cabin` will be filled with `"Unknown"` because it contains a large proportion of missing values, and this approach allows us to preserve the passenger records while explicitly representing the unavailable cabin information.

This strategy helps preserve the dataset while handling missing values appropriately according to the type and amount of missing data.

```python
# Handle missing values

# Fill missing Age values with the median
df["Age"] = df["Age"].fillna(df["Age"].median())

# Fill missing Embarked values with the mode
df["Embarked"] = df["Embarked"].fillna(df["Embarked"].mode()[0])

# Fill missing Cabin values with "Unknown"
df["Cabin"] = df["Cabin"].fillna("Unknown")

print("Missing values have been handled successfully.")
```

```python
# Verify missing values after cleaning

missing_values_after = df.isnull().sum()

print("Missing Values After Cleaning:")
print(missing_values_after)
```

### Missing Value Handling Justification

The `fillna()` method was selected instead of `dropna()` because removing rows with missing values could unnecessarily reduce the size of the Titanic dataset.

- **Age:** Missing values were replaced with the median age because `Age` is a numerical feature, and the median is less affected by extreme values.
- **Embarked:** Missing values were replaced with the mode because `Embarked` is a categorical feature and only a small number of values were missing.
- **Cabin:** Missing values were replaced with `"Unknown"` because the column contains a large proportion of missing values. This approach preserves passenger records while indicating that cabin information was unavailable.

After applying these treatments, the dataset was verified and no missing values remained.

### Survival by Gender

Gender is an important feature to investigate because survival outcomes may differ between male and female passengers. The survival rates for each gender will be calculated and compared to identify whether gender shows a noticeable relationship with passenger survival.

```python
# Compare survival counts by gender

survival_by_gender = pd.crosstab(
    df["Sex"],
    df["Survived"]
)

survival_by_gender.columns = ["Did Not Survive", "Survived"]

print("Survival Counts by Gender:")
display(survival_by_gender)
```

```python
# Calculate survival rates by gender

survival_rate_by_gender = df.groupby("Sex")["Survived"].mean() * 100

print("Survival Rate by Gender:")
display(survival_rate_by_gender.round(2))
```

### Observation

The survival comparison shows a clear difference between genders. Female passengers had a substantially higher survival rate than male passengers in the Titanic dataset. This indicates that `Sex` has a strong relationship with survival and should be considered an important feature in the subsequent analysis.

### Survival by Passenger Class

Passenger class is another important feature to investigate because passengers were divided into three classes on the Titanic. The survival outcomes and survival rates for each passenger class will be compared to determine whether passenger class has a noticeable relationship with survival.

```python
# Compare survival counts by passenger class

survival_by_class = pd.crosstab(
    df["Pclass"],
    df["Survived"]
)

survival_by_class.columns = ["Did Not Survive", "Survived"]

print("Survival Counts by Passenger Class:")
display(survival_by_class)
```

```python
# Calculate survival rates by passenger class

survival_rate_by_class = df.groupby("Pclass")["Survived"].mean() * 100

print("Survival Rate by Passenger Class:")
display(survival_rate_by_class.round(2))
```

### Observation

The survival comparison shows noticeable differences between passenger classes. First-class passengers had a higher survival rate than second- and third-class passengers, while third-class passengers had the lowest survival rate. This indicates that `Pclass` has a strong relationship with survival and is an important feature for further analysis.

### Investigation of Other Relevant Features

In addition to gender and passenger class, other numerical features such as age and fare may also influence passenger survival. This analysis compares the average age and average fare of passengers based on their survival status to identify any noticeable patterns.

```python
# Compare average age by survival status

average_age = df.groupby("Survived")["Age"].mean()

print("Average Age by Survival Status:")
display(average_age.round(2))
```

```python
# Compare average fare by survival status

average_fare = df.groupby("Survived")["Fare"].mean()

print("Average Fare by Survival Status:")
display(average_fare.round(2))
```

```python
# Summary of selected numerical features by survival

feature_summary = df.groupby("Survived")[["Age", "Fare"]].mean()

print("Summary of Numerical Features by Survival:")
display(feature_summary.round(2))
```

### Observation

The analysis shows noticeable differences in the average fare and average age of passengers based on survival status. Passengers who survived generally paid higher fares, suggesting that fare may be associated with a higher likelihood of survival. The difference in average age is comparatively smaller, indicating that age may have had a less pronounced relationship with survival than features such as gender, passenger class, and fare.
