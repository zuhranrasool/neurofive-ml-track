```python
# Import required libraries

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
# Load the Titanic dataset

df = pd.read_csv("../data/train.csv")

print("Dataset loaded successfully!")
```

```python
# Display the first five rows of the dataset

df.head()
```

```python
# Display the number of rows and columns

rows, columns = df.shape

print(f"Number of rows: {rows}")
print(f"Number of columns: {columns}")
```

```python
# Display dataset information

df.info()
```

```python
# Display summary statistics for numerical columns

df.describe()
```

```python
# Check missing values in each column

df.isnull().sum()
```

```python
# Check for duplicate records

df.duplicated().sum()
```

```python
# Display data types of all columns

df.dtypes
```

```python
# Display numerical columns

numerical_features = df.select_dtypes(include=['int64', 'float64']).columns

print("Numerical Features:")
print(numerical_features)
```

```python
# Display categorical columns

categorical_features = df.select_dtypes(include=['str']).columns

print("Categorical Features:")
print(categorical_features)
```

```python
# Missing Values Overview

missing_values = df.isnull().sum()

plt.figure(figsize=(10, 5))

missing_values.plot(kind='bar')

plt.title("Missing Values in Titanic Dataset")
plt.xlabel("Columns")
plt.ylabel("Number of Missing Values")

plt.xticks(rotation=45)

plt.tight_layout()

plt.savefig("../images/missing_values.png", dpi=300, bbox_inches="tight")

plt.show()
```

```python
# Passenger Survival Distribution

plt.figure(figsize=(6, 5))

sns.countplot(x='Survived', data=df)

plt.title("Passenger Survival Distribution")
plt.xlabel("Survival Status")
plt.ylabel("Number of Passengers")

plt.xticks([0, 1], ['Did Not Survive', 'Survived'])

plt.tight_layout()

plt.savefig("../images/survival_distribution.png", dpi=300, bbox_inches="tight")

plt.show()
```

```python
# Gender Distribution

plt.figure(figsize=(6, 5))

sns.countplot(x='Sex', data=df)

plt.title("Gender Distribution of Passengers")
plt.xlabel("Gender")
plt.ylabel("Number of Passengers")

plt.tight_layout()

plt.savefig("../images/gender_distribution.png", dpi=300, bbox_inches="tight")

plt.show()
```

```python
# Passenger Class Distribution

plt.figure(figsize=(6, 5))

sns.countplot(x='Pclass', data=df)

plt.title("Passenger Class Distribution")
plt.xlabel("Passenger Class")
plt.ylabel("Number of Passengers")

plt.xticks([0, 1, 2], ['First Class', 'Second Class', 'Third Class'])

plt.tight_layout()

plt.savefig("../images/passenger_class_distribution.png", dpi=300, bbox_inches="tight")

plt.show()
```

```python
# Age Distribution

plt.figure(figsize=(8, 5))

sns.histplot(df['Age'], bins=30, kde=True)

plt.title("Age Distribution of Titanic Passengers")
plt.xlabel("Age")
plt.ylabel("Number of Passengers")

plt.tight_layout()

plt.savefig("../images/age_distribution.png", dpi=300, bbox_inches="tight")

plt.show()
```

```python
# Fare Distribution

plt.figure(figsize=(8, 5))

sns.histplot(df['Fare'], bins=30, kde=True)

plt.title("Fare Distribution of Titanic Passengers")
plt.xlabel("Fare")
plt.ylabel("Number of Passengers")

plt.tight_layout()

plt.savefig("../images/fare_distribution.png", dpi=300, bbox_inches="tight")

plt.show()
```

```python
# Correlation Heatmap

# Select only numerical columns
numerical_df = df.select_dtypes(include=['int64', 'float64'])

# Create correlation matrix
correlation_matrix = numerical_df.corr()

# Plot heatmap
plt.figure(figsize=(10, 8))

sns.heatmap(
    correlation_matrix,
    annot=True,
    cmap="coolwarm",
    fmt=".2f",
    linewidths=0.5
)

plt.title("Correlation Heatmap of Titanic Dataset")

plt.tight_layout()

plt.savefig("../images/correlation_heatmap.png", dpi=300, bbox_inches="tight")

plt.show()
```

```python
# Survival by Gender

plt.figure(figsize=(7, 5))

sns.countplot(x='Sex', hue='Survived', data=df)

plt.title("Survival by Gender")
plt.xlabel("Gender")
plt.ylabel("Number of Passengers")

plt.legend(title="Survived", labels=["No", "Yes"])

plt.tight_layout()

plt.savefig("../images/survival_by_gender.png", dpi=300, bbox_inches="tight")

plt.show()
```

```python
# Survival by Passenger Class

plt.figure(figsize=(8, 5))

sns.countplot(x='Pclass', hue='Survived', data=df)

plt.title("Survival by Passenger Class")
plt.xlabel("Passenger Class")
plt.ylabel("Number of Passengers")

plt.xticks([0, 1, 2], ["First Class", "Second Class", "Third Class"])

plt.legend(title="Survived", labels=["No", "Yes"])

plt.tight_layout()

plt.savefig("../images/survival_by_class.png", dpi=300, bbox_inches="tight")

plt.show()
```

# Phase 9: Data Story

## Data Story

The Titanic dataset contains **891 passenger records** and **12 features**, providing demographic and travel-related information for each passenger.

During the exploratory data analysis, missing values were identified primarily in the **Cabin**, **Age**, and **Embarked** columns, while the remaining features were complete.

The dataset includes both **numerical features** (such as Age, Fare, SibSp, Parch, and Passenger Class) and **categorical features** (such as Sex, Embarked, Cabin, Ticket, and Name).

Visual analysis showed that **Third Class** had the highest number of passengers, while **First Class** passengers experienced a higher survival rate.

The survival analysis also revealed that **female passengers survived at a significantly higher rate than male passengers**, making gender one of the most influential features in the dataset.

Overall, the exploratory data analysis provides valuable insights into passenger demographics, survival patterns, and feature relationships, establishing a strong foundation for the machine learning phase of the project.
